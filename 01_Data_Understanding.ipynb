# **Data Understanding**

This notebook loads the e-commerce dataset and inspects its structure, data types, summary statistics, missing values, and duplicate records.

```python
# Load pandas for data manipulation.
import pandas as pd

# Set option to display all columns and format float values to two decimal places.
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:,.2f}'.format)
```

```python
# Read the raw e-commerce dataset.
ecommerce_df = pd.read_csv(r'dataset/raw/ecommerce_dataset_+1m.csv')
```

## 1. Look at the Data

```python
# Display a random sample of 10 rows
ecommerce_df.sample(10)
```

```python
# Dataset dimensions (rows, columns)
ecommerce_df.shape
```

## 2. Data Types & Structure

```python
# Concise summary of column types and non-null counts
ecommerce_df.info()
```

## 3. Descriptive Statistics

```python
# Summary statistics for numerical columns
ecommerce_df.describe()
```

## 4. Missing Values

```python
# Percentage of missing values per column (%)
print((ecommerce_df.isnull().sum() / ecommerce_df.shape[0] * 100).to_string())
```

**Notable missing values:**
- `return_reason` → ~90.01% missing (expected, only present for returned orders)
- `customer_feedback` → ~19.96% missing
- `coupon_code` → ~50.00% missing (expected, only present when a coupon was used)

## 5. Duplicate Check

```python
# Total number of fully duplicate rows
duplicates_row = ecommerce_df.duplicated().sum()
print(f'Total Duplicate Rows : {duplicates_row}')
```

## 6. Unique Identifier Check

```python
"""
    Check which columns have completely unique values
    True = potential unique identifier
 """
print(ecommerce_df.apply(lambda col: col.is_unique).to_string())
```

> **Observation:**  
> The dataset does **not** have a reliable unique identifier column.  
> Although `order_date` appears unique in this sample, timestamps are not dependable as primary keys in real-world scenarios (collisions can occur).
