# Data Understanding
This notebook loads the e-commerce dataset and inspects its structure, data types, summary statistics, missing values, and duplicate records.

```python
# Load pandas for data manipulation.
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:,.2f}'.format)
```

```python
# Read the raw e-commerce dataset from the project dataset folder.
ecommerce_df = pd.read_csv(r'dataset/raw/ecommerce_dataset_+1m.csv')
```

```python
# DIsplay a random sample of 10 rows from the dataset to get an overview of the data.
ecommerce_df.sample(10)
```

```python
# Check the overall dataset dimensions: rows and columns.
ecommerce_df.shape
```

```python
# Print a concise summary of column types and non-null counts.
ecommerce_df.info()
```

```python
# Compute descriptive statistics
ecommerce_df.describe()
```

```python
# Calculate and print the percentage of missing values per column.
print((ecommerce_df.isnull().sum() / ecommerce_df.shape[0] * 100).round(2).to_string())
```

Notable missing values were observed in **``return_season``**, **``customer_feedback``**, and **``coupon_code``**.

```python
# Total of duplicate rows
duplicates_row = ecommerce_df.duplicated().sum()
print(f'Total of Duplicate Rows : {duplicates_row}')
```

```python
# Number of duplicate values per column
duplicates_per_col = ecommerce_df.apply(lambda col: col.duplicated().sum())
print(duplicates_per_col.to_string())
```

>Based on the data, it appears that the dataset does not have a unique identifier column.
